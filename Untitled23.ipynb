```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
url = 'https://breathecode.herokuapp.com/asset/internal-link?id=927&path=AB_NYC_2019.csv'
df = pd.read_csv(url)
df
```

```python
print('Dimensiones del dataset:', df.shape)
df.head()
```

```python
# Inspecciónde tipos de datos y valores nulos
print('---Información del dataset---')
df.info()
print('\n---Valores faltantes por columna---')
print(df.isnull().sum())
```

```python
# Eliminación de columnas irrelevantes para modelado/análisis estadístico
# 'id', 'name', 'host_id, 'host_name' no aportan valor predictivo directo
cols_to_drop = ['id', 'name', 'host_id', 'host_name', 'last_review']
df_clean = df.drop(columns=cols_to_drop)
# Imputación de nulos en 'reviews_per_month' (si no tiene reviews, el valor es 0)
df_clean['reviews_per_month'] = df_clean['reviews_per_month'].fillna(0)
print('Nulos restantes después de la limpieza:')
print(df_clean.isnull().sum())
```

Conclusión:

Se eliminan las identificaciones únicas e información textual (id, name, hot_id, host_name) al no aportar señal predictiva para el precio o comportamiento del hospedaje. La columna last_review presenta un número alto de ausentes por publicaciones sin reseñas registradas; en tanto, para reviews_per_month se asume un valor de 0 cuando no existe historial previo.

```python
# Análisis univariado y detección de outliersles
# Estadísticas descriptivas de variables numéricas
print('--- Estadísticas descriptivas ---')
print(df_clean.describe())
# Visualización de la variable objetivo (price) y destribución por tipo de habitación
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(df_clean['price'], bins=50, kde=True, color='teal')
plt.title('Distribución del precio')
plt.xlabel('Precio ($)')

plt.subplot(1, 2, 2)
sns.boxplot(
    x='room_type', y='price', data=df_clean[df_clean['price'] < 500], palette='Set2'
    )
plt.title('Precio según tipo de habitación (< $500)')
plt.xlabel('Tipo de habitación')
plt.ylabel('precio ($)')

plt.tight_layout()
plt.show()
```

Conclusión:

La variable price presenta un sesgo positivo pronunciado con la presencia de *outliers* severos (valores máximos alcanzando lo $10.000 USD, mientras que la mediana se ubica en torno a $106 USD). Los departamentos completos (Entire home/apt) registran la mayor variabilidad y precio medio, mientras que las habitaciones compartidas (Share room) representan el segmento más económico.

```python
# Análisis bivariado y geográfico
# Matriz de correlación entre variables numéricas
plt.figure(figsize=(8, 6))
sns.heatmap(
    df_clean.select_dtypes(include=[np.number]).corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1,
)
plt.title('Matriz de correlación')
plt.show()

# Distribución geográfica según sector (neighbourhood_group) y precio
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_clean[df_clean['price'] < 500],
    x='longitude',
    y='latitude',
    hue='neighbourhood_group',
    size='price',
    sizes=(10, 200),
    alpha=0.5,
    palette='viridis',
)
plt.title('distribución geográfica por distrito y rango de precio')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

Conclusión:

La ubicación geográfica determina fuertemente la tarifa. Manhattan concentra la mayor densidad de ofertas con precios superiores, seguido de Brooklyn. En contraste, sectores como el Bronx y Staten Island mantienen tarifas significativamente menores. En la matriz de correlación no se observan relaciones lineales fuertes con la variables objetivo, lo que sugiere una estructura no lineal dependiente de variables categóricas (distrito y tipo de habitación).

```python
# División Train/Test y guardado del dataset
# Codoficación rápida de variables categóricas esenciales para la división
df_processed = pd.get_dummies(
    df_clean, columns=['neighbourhood_group', 'room_type'], drop_first=True
)
# Separa en Train (80%) y Test (20%)
train_df, test_df = train_test_split(
    df_processed, test_size=0.2, random_state=42
)

print(f'Tamaño conjunto Train: {train_df.shape}')
print(f'Tamaño conjunto Test: {test_df.shape}')

#Crear directorios y guardar archivos según la estructura requerida
os.makedirs('data/raw', exist_ok=True)
os.makedirs('data/processed', exist_ok=True)

# Guardar
df.to_csv('data/raw/AB_NYC_2019.csv', index=False)
train_df.to_csv('data/processed/train.csv', index=False)
test_df.to_csv('data/processed/test.csv', index=False)

print("\nArchivos procesados gradados correctamente en 'data/processer/'.")
```
